```python
import os
import shutil

# Define paths
kaggle_dir = os.path.expanduser('~/.kaggle')
source_path = '/content/drive/MyDrive/major_project/kaggle.json'
destination_path = os.path.join(kaggle_dir, 'kaggle.json')

# Create .kaggle directory if it doesn't exist
os.makedirs(kaggle_dir, exist_ok=True)

# Copy kaggle.json from Google Drive to local session
if os.path.exists(source_path):
    shutil.copy(source_path, destination_path)
    # Set permissions to 600 (owner read/write only)
    os.chmod(destination_path, 0o600)
    print(f'Successfully configured Kaggle API. File copied to {destination_path}')
else:
    print(f'Error: kaggle.json not found at {source_path}')

# Verify the file existence
if os.path.exists(destination_path):
    print('Verification: kaggle.json is present in the .kaggle directory.')
```

```python
import os

# Define the base dataset path
dataset_root = '/content/drive/MyDrive/major_project/military_object_dataset'

# 1. List main directories
main_dirs = os.listdir(dataset_root)
print(f'Main directories in {dataset_root}: {main_dirs}')

# 2 & 3. Explore train/val and count files
for split in ['train', 'val']:
    split_path = os.path.join(dataset_root, split)
    if os.path.exists(split_path):
        subfolders = os.listdir(split_path)
        print(f'\nSubdirectories in {split}: {subfolders}')

        img_path = os.path.join(split_path, 'images')
        if os.path.exists(img_path):
            images = os.listdir(img_path)
            print(f'Number of images in {split}/images: {len(images)}')
            # 4. Print sample file path
            if len(images) > 0:
                print(f'Sample image path: {os.path.join(img_path, images[0])}')
        else:
            print(f'Warning: {split}/images directory not found.')
    else:
        print(f'Warning: {split} directory not found.')
```

```python
from ultralytics import YOLO
import os

# Define paths
weights_path = '/content/drive/MyDrive/major_project/best.pt'
yaml_config_path = '/content/drive/MyDrive/major_project/military_dataset.yaml'

# 3. Instantiate the YOLO model
if os.path.exists(weights_path):
    model = YOLO(weights_path)
    print(f'Successfully loaded YOLO model from {weights_path}')

    # 4. Perform model validation
    print('Starting validation...')
    results = model.val(data=yaml_config_path)

    # 5. Print/display validation results
    print('\nValidation Results:')
    print(f'Mean Average Precision @.5:.95 (mAP50-95): {results.box.map:.4f}')
    print(f'Mean Average Precision @.5 (mAP50): {results.box.map50:.4f}')
else:
    print(f'Error: Model weights not found at {weights_path}')
```

```python
import os
from ultralytics import YOLO

# List files in the project directory to find the correct weights
project_dir = '/content/drive/MyDrive/major_project/'
files = os.listdir(project_dir)
print(f'Files in {project_dir}: {files}')

# Filter for .pt files
pt_files = [f for f in files if f.endswith('.pt')]
print(f'Detected weight files: {pt_files}')

# Try to use 'v8best.pt' as a fallback if 'best.pt' is missing, as it is common in YOLO runs
fallback_weights = os.path.join(project_dir, 'v8best.pt')
yaml_config_path = '/content/drive/MyDrive/major_project/military_dataset.yaml'

if os.path.exists(fallback_weights):
    print(f'Attempting validation with: {fallback_weights}')
    model = YOLO(fallback_weights)
    results = model.val(data=yaml_config_path)
    print('\nValidation Results:')
    print(f'mAP50-95: {results.box.map:.4f}')
    print(f'mAP50: {results.box.map50:.4f}')
else:
    print('Could not find a valid weights file to proceed with validation.')
```
