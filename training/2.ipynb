```python
import pickle, sys, runpy
from pathlib import PosixPath

# Patch Unpickler to replace WindowsPath with PosixPath
class CustomUnpickler(pickle.Unpickler):
    def find_class(self, module, name):
        if module == "pathlib" and name == "WindowsPath":
            return PosixPath
        return super().find_class(module, name)

pickle.Unpickler = CustomUnpickler
print("✅ Patched WindowsPath → PosixPath")

# Arguments for val.py
val_args = [
    "val.py",
    "--weights", "/content/drive/MyDrive/major_project/v5best.pt",
    "--data", "/content/military_dataset.yaml",
    "--batch-size", "8",
    "--imgsz", "640"
]

# Run val.py with patch active
original_sys_argv = sys.argv
sys.argv = val_args
try:
    runpy.run_path("val.py", run_name="__main__")
finally:
    sys.argv = original_sys_argv
```
